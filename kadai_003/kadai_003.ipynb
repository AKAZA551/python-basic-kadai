```python
# step1
import numpy as np

# step2
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[0, 1],
              [2, 3],
              [4, 5]])

# step3
result_dot = np.dot(A, B)

# step4
print("行列の積:")
print(result_dot)

# step5
max_value = np.max(result_dot)
print("要素の最大値:", max_value)
```
