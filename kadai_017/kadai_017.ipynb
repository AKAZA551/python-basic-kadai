```python
class Human:

  def __init__(self,name :str,age :int):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}さんは大人です。")
    else:
      print(f"{self.name}さんは大人ではありません。")

humans = [Human("五条 悟", 29),Human("東方 仗助", 16)]

for human in humans:
    human.check_adult()
```
